```python
import pandas as pd

df = pd.read_csv(
    "Ecommerce _Customer_Data.csv",
    encoding="latin1"
)

print(df.shape)
```

```python
reference_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)

reference_date
```

```python
df.columns
```

```python
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    format="mixed"
)
```

```python
df["InvoiceDate"].dtype
```

```python
reference_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)

reference_date
```

```python
rfm = df.groupby("CustomerID").agg({
    "InvoiceDate": lambda x: (reference_date - x.max()).days,
    "InvoiceNo": "nunique",
    "TotalAmount": "sum"
})

rfm.columns = ["Recency", "Frequency", "Monetary"]
```

```python
rfm.head()
```

```python
df["TotalAmount"] = df["Quantity"] * df["UnitPrice"]
```

```python
df[["Quantity", "UnitPrice", "TotalAmount"]].head()
```

```python
rfm = df.groupby("CustomerID").agg({
    "InvoiceDate": lambda x: (reference_date - x.max()).days,
    "InvoiceNo": "nunique",
    "TotalAmount": "sum"
})

rfm.columns = ["Recency", "Frequency", "Monetary"]
```

```python
rfm.head()
```

```python
rfm.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(rfm)
```

```python
rfm_scaled[:5]
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []
K = range(2, 11)

for k in K:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(rfm_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K, inertia, marker="o")
plt.title("Elbow Method for Optimal K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(K)
plt.grid(True)
plt.show()
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

rfm["Cluster"] = kmeans.fit_predict(rfm_scaled)
```

```python
rfm.head()
```

```python
rfm["Cluster"].value_counts().sort_index()
```

```python
cluster_summary = rfm.groupby("Cluster").agg({
    "Recency": "mean",
    "Frequency": "mean",
    "Monetary": "mean"
}).round(2)

cluster_summary
```

```python
cluster_summary["Customer_Count"] = rfm["Cluster"].value_counts()

cluster_summary
```

```python
import matplotlib.pyplot as plt

cluster_counts = rfm["Cluster"].value_counts().sort_index()

plt.figure(figsize=(8, 5))
plt.bar(
    cluster_counts.index.astype(str),
    cluster_counts.values
)

plt.title("Customer Distribution by Cluster")
plt.xlabel("Customer Cluster")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
cluster_summary[["Recency", "Frequency", "Monetary"]].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("RFM Comparison by Customer Cluster")
plt.xlabel("Customer Cluster")
plt.ylabel("Average Value")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
cluster_summary
```

```python
segment_names = {
    0: "Regular Customers",
    1: "At-Risk / Low-Value",
    2: "VIP / Champions",
    3: "High-Value Loyal"
}

rfm["Segment"] = rfm["Cluster"].map(segment_names)
```

```python
rfm.head()
```

```python
final_summary = rfm.groupby("Segment").agg({
    "Recency": "mean",
    "Frequency": "mean",
    "Monetary": "mean"
}).round(2)

final_summary
```

```python
rfm.to_csv("customer_segmentation_final.csv")

final_summary.to_csv("customer_segmentation_summary.csv")
```
